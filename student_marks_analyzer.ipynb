# 📊 Student Marks Analysis using NumPy

**Objective:** Generate marks of 50 students in 5 subjects and perform comprehensive analysis including toppers, failures, subject-wise averages, and grade bands.

### NumPy Topics Covered:
- `np.array`
- `np.random.randint`
- `shape`, `ndim`, `size`, `dtype` inspection
- Basic indexing and negative indexing
- 2D slicing (rows, columns, sub-matrices)
- Boolean masking for conditional analysis

---

## 🔹 Step 1: Create Subject Names and Marks Matrix
- Create subject names array using `np.array()`
- Generate random marks between 35 and 100 for 50 students across 5 subjects with `np.random.randint(35, 100, (50, 5))`

```python
import numpy as np

# Set random seed for reproducible results
np.random.seed(42)

# 1. Define subject names using np.array
subjects = np.array(['Mathematics', 'Physics', 'Chemistry', 'English', 'Computer Science'])

# 2. Generate random integer marks for 50 students across 5 subjects
# Range: [35, 100) -> Minimum 35, Maximum 99
marks = np.random.randint(35, 100, size=(50, 5))

print("Subjects:", subjects)
print("\nFirst 5 Students' Marks Matrix:")
print(marks[:5])
```

## 🔹 Step 2: Inspect Array Attributes
Check `shape`, `ndim`, `size`, and `dtype` of the marks array.

```python
print("=" * 45)
print("ARRAY ATTRIBUTES SUMMARY")
print("=" * 45)
print(f"• Shape (Dimensions) : {marks.shape} -> ({marks.shape[0]} students, {marks.shape[1]} subjects)")
print(f"• Ndim (Axes count)  : {marks.ndim} -> 2D matrix")
print(f"• Size (Total items) : {marks.size} -> 50 x 5 = 250 individual scores")
print(f"• Dtype (Data type)  : {marks.dtype} -> 64-bit integer")
print("=" * 45)
```

## 🔹 Step 3: Indexing and Slicing Operations
Explore basic indexing, negative indexing, and 2D slicing:
1. First 10 students
2. Last subject (negative indexing `[:, -1]`)
3. Column ranges (e.g. STEM subjects `[:, 0:3]`)
4. Row and column sub-matrix slices

```python
# 1. Single element: Student 0 in Chemistry (Column index 2)
print(f"1. Student 0's Chemistry mark: {marks[0, 2]}")

# 2. First 10 students (marks[:10, :])
print("\n2. First 10 students (all subjects):")
print(marks[:10])

# 3. Last subject using negative index (marks[:, -1])
print(f"\n3. Last subject ('{subjects[-1]}') for first 5 students:")
print(marks[:5, -1])

# 4. Column range: Science subjects (Maths, Physics, Chemistry -> columns 0 to 2)
print(f"\n4. Science subjects ({subjects[0:3]}) for first 5 students:")
print(marks[:5, 0:3])

# 5. Slicing rows & columns: Students 15 to 19 in Physics & Chemistry
print(f"\n5. Students 15 to 19 in {subjects[1:3]} [marks[15:20, 1:3]]:")
print(marks[15:20, 1:3])
```

## 🔹 Step 4: Boolean Masking & Grade Bands
Apply boolean masks to identify:
- Scores below 40 (Failures)
- Scores above 90 (Distinctions)
- Student classification across Grade Bands (A+, A, B, C, D, F)

```python
# 1. Failures: Marks below 40
fail_mask = marks < 40
print(f"Total failing exam instances (< 40): {np.sum(fail_mask)}")

# Find students who failed in at least one subject
students_with_fail = np.any(fail_mask, axis=1)
failed_student_ids = np.where(students_with_fail)[0]
print(f"Number of students who failed in at least 1 subject: {len(failed_student_ids)}")
print(f"Failed Student IDs: {failed_student_ids}")

# 2. Distinctions: Marks above 90
distinction_mask = marks > 90
print(f"\nTotal distinction marks (> 90): {np.sum(distinction_mask)}")
students_with_distinction = np.any(distinction_mask, axis=1)
print(f"Students with at least one distinction: {np.sum(students_with_distinction)}")

# 3. Student-level averages and Grade Bands
student_averages = np.mean(marks, axis=1)

grade_bands = {
    'A+ (>= 90)'   : student_averages >= 90,
    'A  (80 - 89)' : (student_averages >= 80) & (student_averages < 90),
    'B  (70 - 79)' : (student_averages >= 70) & (student_averages < 80),
    'C  (60 - 69)' : (student_averages >= 60) & (student_averages < 70),
    'D  (40 - 59)' : (student_averages >= 40) & (student_averages < 60),
    'F  (< 40)'    : student_averages < 40
}

print("\n--- GRADE BAND DISTRIBUTION (Based on Student Average) ---")
for grade, mask in grade_bands.items():
    count = np.sum(mask)
    pct = (count / len(marks)) * 100
    print(f"Grade {grade:12s} : {count:2d} students ({pct:5.1f}%)")
```

## 🔹 Step 5: Subject-wise Toppers, Averages & Pass Percentages
- Calculate subject-wise averages using `np.mean(marks, axis=0)`
- Find subject-wise maximum marks and identify topper(s) using boolean masking / `np.argmax`
- Compute pass percentage per subject using boolean mask `marks >= 40`
- Determine overall class topper and all-subject pass percentage

```python
subject_averages = np.mean(marks, axis=0)
subject_max = np.max(marks, axis=0)
pass_mask = marks >= 40
subject_pass_counts = np.sum(pass_mask, axis=0)
subject_pass_pct = (subject_pass_counts / len(marks)) * 100

print("=" * 75)
print(f"{'Subject':<18} | {'Average':>7} | {'Max':>4} | {'Topper ID(s)':<14} | {'Pass %':>7} | {'Passed/Total'}")
print("=" * 75)

for i, subj in enumerate(subjects):
    col = marks[:, i]
    toppers = np.where(col == subject_max[i])[0]
    topper_str = ", ".join(f"#{t}" for t in toppers)
    print(f"{subj:<18} | {subject_averages[i]:7.2f} | {subject_max[i]:4d} | {topper_str:<14} | {subject_pass_pct[i]:6.1f}% | {subject_pass_counts[i]}/{len(marks)}")

print("=" * 75)

# Overall Class Topper (Highest aggregate score)
totals = np.sum(marks, axis=1)
overall_topper = np.argmax(totals)
print(f"\n🏆 Overall Class Topper: Student #{overall_topper}")
print(f"   Total Marks: {totals[overall_topper]} / 500 (Average: {student_averages[overall_topper]:.2f}%)")
print(f"   Score per subject: {dict(zip(subjects, marks[overall_topper]))}")

# Students who passed ALL 5 subjects
passed_all = np.all(marks >= 40, axis=1)
print(f"\n🎓 Students Passing ALL 5 Subjects: {np.sum(passed_all)} / {len(marks)} ({(np.sum(passed_all)/len(marks))*100:.1f}%)")
```
